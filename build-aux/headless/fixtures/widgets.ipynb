# Interactive widgets

Drawn as GTK widgets, not a web page. Saved with their state, so they show before a kernel runs; running a cell makes them live.

```python
import ipywidgets as w
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
rng = np.random.default_rng(3)
df = pd.DataFrame({"x": rng.normal(size=300), "g": rng.choice(["a", "b", "c"], 300)})
```

```python
@w.interact(bins=(5, 40), kind=["hist", "kde"])
def show(bins=20, kind="hist"):
    if kind == "hist":
        sns.histplot(df, x="x", hue="g", bins=bins)
    else:
        sns.kdeplot(df, x="x", hue="g")
    plt.show()
```

```python
w.VBox([
    w.FloatLogSlider(value=10, base=10, min=-2, max=3, description="Rate"),
    w.Dropdown(options=["linear", "log", "symlog"], value="log", description="Scale"),
    w.Checkbox(value=True, description="Show grid"),
    w.Text(value="run-7", description="Label"),
    w.ToggleButtons(options=["train", "eval", "test"], description="Split"),
    w.IntProgress(value=7, max=10, description="Epochs"),
    w.HBox([w.Button(description="Run", button_style="primary"), w.Button(description="Stop", button_style="danger")]),
])
```

```python
w.Tab(children=[w.RadioButtons(options=["mean", "median", "mode"], description="Statistic"), w.ColorPicker(value="#3584e4", description="Colour")], titles=["Summary", "Style"])
```
