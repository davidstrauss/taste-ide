# Figures from matplotlib and Seaborn

Each figure below came back from a real kernel through the IDE's bridge: a default PNG, a **retina** PNG drawn at twice its size, an **SVG**, and Seaborn's objects interface.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
rng = np.random.default_rng(7)
df = pd.DataFrame({"x": rng.normal(size=200), "y": rng.normal(size=200)})
df["group"] = np.where(df.x > 0, "right", "left")
print(df.shape)
```

```python
fig, ax = plt.subplots(figsize=(6, 2.5))
ax.plot(np.sort(df.x), marker=".", linestyle="none")
ax.set_title("matplotlib, PNG")
```

```python
%config InlineBackend.figure_format = 'retina'
sns.set_theme(style="whitegrid")
sns.histplot(df, x="x", hue="group", bins=20).set_title("Seaborn, retina PNG")
```

```python
%config InlineBackend.figure_formats = ['svg']
sns.scatterplot(df, x="x", y="y", hue="group").set_title("Seaborn, SVG")
```

```python
%config InlineBackend.figure_formats = ['png']
import seaborn.objects as so
so.Plot(df, x="x", y="y", color="group").add(so.Dot()).label(title="Seaborn objects")
```
