# Squares, sampled

A short notebook for looking at the IDE's notebook face: prose, code, its **outputs**, and a `traceback`.

- a stream
- a result
- a figure


```python
xs = list(range(12))
print(f"sampling {len(xs)} points")
squares = [x * x for x in xs]
print("done")
```

```python
squares[-1]
```

```python
import matplotlib.pyplot as plt
fig, ax = plt.subplots(figsize=(6, 3))
ax.plot(xs, squares, marker="o")
ax.set_title("Squares")
```

A DataFrame shows its text beside its HTML, and the text is what is drawn:

```python
import pandas as pd
pd.DataFrame({"x": xs[:4], "square": squares[:4]})
```
